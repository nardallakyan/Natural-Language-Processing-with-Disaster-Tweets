```python
!pip install -q transformers datasets accelerate scikit-learn pandas numpy torch

import os
import gc
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score
from datasets import Dataset
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    Trainer,
    TrainingArguments,
    DataCollatorWithPadding
)

def seed_everything(seed=42):
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.backends.cudnn.deterministic = True

seed_everything(42)
```

```python

train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')

def preprocess_data(df):
    df = df.copy()
    df['keyword'] = df['keyword'].fillna('none').astype(str)
    df['location'] = df['location'].fillna('none').astype(str)
    df['text'] = df['text'].fillna('').astype(str)

    
    df['full_text'] = (
        "Keyword: " + df['keyword'] + 
        " | Location: " + df['location'] + 
        " | Tweet: " + df['text']
    )
    return df

train_df = preprocess_data(train_df)
test_df = preprocess_data(test_df)

print(f"Train set shape: {train_df.shape}")
print(f"Test set shape: {test_df.shape}")
print("\nSample Processed Input:")
print(train_df['full_text'].iloc[0])
```

```python
MODEL_NAME = "microsoft/deberta-v3-base"
MAX_LEN = 160

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

def tokenize_fn(examples):
    return tokenizer(
        examples["full_text"],
        truncation=True,
        max_length=MAX_LEN,
        padding=False  
    )
```

```python
N_SPLITS = 5
skf = StratifiedKFold(n_splits=N_SPLITS, shuffle=True, random_state=42)

oof_preds = np.zeros(len(train_df))
test_preds = np.zeros(len(test_df))


ds_test = Dataset.from_pandas(test_df[['full_text']])
ds_test = ds_test.map(tokenize_fn, batched=True)
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

for fold, (train_idx, val_idx) in enumerate(skf.split(train_df, train_df['target'])):
    print(f"\n{'='*20} Fold {fold + 1} / {N_SPLITS} {'='*20}")
    
    fold_train_df = train_df.iloc[train_idx].reset_index(drop=True)
    fold_val_df = train_df.iloc[val_idx].reset_index(drop=True)
    
    ds_train = Dataset.from_pandas(fold_train_df[['full_text', 'target']])
    ds_val = Dataset.from_pandas(fold_val_df[['full_text', 'target']])
    
    ds_train = ds_train.map(tokenize_fn, batched=True)
    ds_val = ds_val.map(tokenize_fn, batched=True)
    
    model = AutoModelForSequenceClassification.from_pretrained(
        MODEL_NAME,
        num_labels=2
    )
    
    training_args = TrainingArguments(
        output_dir=f"./results_fold_{fold}",
        eval_strategy="epoch",
        save_strategy="epoch",
        learning_rate=2e-5,
        per_device_train_batch_size=16,
        per_device_eval_batch_size=32,
        num_train_epochs=3,
        weight_decay=0.01,
        warmup_ratio=0.1,
        fp16=torch.cuda.is_available(),
        logging_steps=50,
        load_best_model_at_end=True,
        metric_for_best_model="f1",
        greater_is_better=True,
        report_to="none"
    )
    
    def compute_metrics(eval_pred):
        logits, labels = eval_pred
        preds = np.argmax(logits, axis=-1)
        return {"f1": f1_score(labels, preds)}
    
    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=ds_train,
        eval_dataset=ds_val,
        tokenizer=tokenizer,
        data_collator=data_collator,
        compute_metrics=compute_metrics
    )
    
    trainer.train()
    
   
    val_preds_raw = trainer.predict(ds_val).predictions
    val_probs = torch.softmax(torch.tensor(val_preds_raw), dim=-1)[:, 1].numpy()
    oof_preds[val_idx] = val_probs
    
    
    test_preds_raw = trainer.predict(ds_test).predictions
    test_probs = torch.softmax(torch.tensor(test_preds_raw), dim=-1)[:, 1].numpy()
    test_preds += test_probs / N_SPLITS
    
   
    del model, trainer
    torch.cuda.empty_cache()
    gc.collect()
```

```python

best_threshold = 0.5
best_f1 = 0.0

for threshold in np.arange(0.30, 0.70, 0.01):
    score = f1_score(train_df['target'], (oof_preds >= threshold).astype(int))
    if score > best_f1:
        best_f1 = score
        best_threshold = threshold


print(f"Best OOF F1 Score : {best_f1:.4f}")
print(f"Optimal Threshold : {best_threshold:.2f}")


final_preds = (test_preds >= best_threshold).astype(int)


submission_df = pd.DataFrame({
    'id': test_df['id'],
    'target': final_preds
})

submission_df.to_csv('submission.csv', index=False)
print("\n'submission.csv' generated")
print(submission_df.head(10))
```
